# Capstone Evaluation: Comparative Analysis of Classification Models

## Executive Summary
Selecting the optimal classification algorithm requires balancing predictive performance with computational constraints and business requirements. This evaluation compares four fundamental classification algorithms across key criteria and applies them to two distinct domain problems:

1. **Logistic Regression**
2. **Decision Trees**
3. **K-Nearest Neighbors (KNN)**
4. **Support Vector Machines (SVM)**

---
## Criteria Matrix & Trade-Off Analysis

| Model | Imbalanced Data Handling | Training / Inference Speed | Interpretability |
| :--- | :--- | :--- | :--- |
| **Logistic Regression** | Moderate (Supports `class_weight='balanced'`, loss penalty tuning) | **Extremely Fast** ($O(d)$ per iteration, fast inference) | **High** (Direct coefficients, log-odds ratios) |
| **Decision Trees** | Moderate to Good (Can split on minority features, sensitive to depth) | **Fast** ($O(n \cdot d \log n)$ training) | **High** (Visual flowchart, rule sets, feature importance) |
| **KNN** | **Poor** (Biased toward majority class neighbors) | **Lazy Learner** (No training time, slow $O(n \cdot d)$ inference) | **Low to Moderate** (Instance-based, lacks explicit rules) |
| **SVM** | Moderate (Supports class weighting; sensitive to margin shifts) | **Slow** ($O(n^2 \cdot d)$ to $O(n^3)$ for non-linear kernel) | **Low** (Black-box kernel space transformation) |

*Legend:* $n$ = number of samples, $d$ = number of features.

## Task 1 Evaluation: Predicting Customer Churn (`telecom_churn.csv`)

### Context & Requirements
- **Target:** Binary classification (`Churn` = True / False).
- **Key Characteristics:** Typically heavily imbalanced (most customers do not churn). Requires actionable insights for retention strategies.

### Model Analysis & Selection

#### 1. Logistic Regression (Recommended)
- **Pros:** High interpretability through odds ratios (e.g., determining how customer service calls affect churn likelihood). Outputs well-calibrated probabilities allowing threshold tuning for retention interventions. Fast to train and easily adjusted with class weighting.
- **Cons:** Assumes linear decision boundary between log-odds and continuous features.

#### 2. Decision Trees
- **Pros:** Captures non-linear relationships and interactions (e.g., high day minutes combined with customer service calls). Produces clear, rule-based decision logic.
- **Cons:** Prone to overfitting without pruning; sensitive to slight variations in training data.

#### 3. K-Nearest Neighbors (KNN)
- **Pros:** Non-parametric, simple setup.
- **Cons:** Heavily skewed by class imbalance. Distance metrics are sensitive to unscaled numerical features (e.g., account length vs. charges).

#### 4. Support Vector Machines (SVM)
- **Pros:** Effective in high-dimensional spaces with custom kernel functions.
- **Cons:** Hard to explain retention recommendations to business stakeholders. Probability estimates require costly calibration (e.g., Platt scaling).

**Verdict for Task 1:** **Logistic Regression** (or an interpretable **Decision Tree**) is the optimal choice due to executive explainability, ease of probability thresholding, and built-in class balance support.

## Task 2 Evaluation: Recognizing Handwritten Digits (`sklearn.datasets.load_digits`)

### Context & Requirements
- **Target:** Multi-class classification (10 classes: digits 0–9).
- **Key Characteristics:** Pixel-based features ($8 \times 8 = 64$ dimensions). Classes are balanced. Focus is pure accuracy rather than feature explainability.

### Model Analysis & Selection

#### 1. Support Vector Machines (SVM with RBF Kernel - Recommended)
- **Pros:** Superior performance in high-dimensional pixel spaces. Non-linear kernels effectively map complex decision boundaries between handwritten shapes.
- **Cons:** Training time increases with dataset scale, but for standard digit datasets ($8 \times 8$), compute cost is minimal.

#### 2. K-Nearest Neighbors (KNN)
- **Pros:** Intuitive for pixel matching (similar digit images naturally cluster in pixel intensity space). Works well on balanced multi-class tasks.
- **Cons:** Prediction time scales linearly with dataset size; high memory consumption during inference.

#### 3. Decision Trees
- **Pros:** Fast execution.
- **Cons:** Axis-aligned splits perform poorly on subtle rotational and translational spatial image variations.

#### 4. Logistic Regression
- **Pros:** Efficient baseline using One-vs-Rest (OvR) or Multinomial strategy.
- **Cons:** Limited by linear decision boundaries, missing complex pixel co-occurrence relationships.

**Verdict for Task 2:** **SVM (RBF Kernel)** or **KNN** are the preferred choices, prioritizing predictive power and boundary separation over model interpretability.
